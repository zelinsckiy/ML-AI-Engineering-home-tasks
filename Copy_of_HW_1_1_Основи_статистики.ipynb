  3.1. без використання жодних бібліотек, користуючись тільки Python.

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
sorted_data = sorted(sum_purchases)
count = len(sum_purchases)
middle = count // 2

if count % 2 == 1:
  py_median = sorted_data[middle]
else:
  py_median = (sorted_data[middle - 1] + sorted_data [middle]) / 2

print(py_median)
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
x = sum_purchases
y = visits
n = len(x)

mean_x = sum(x) / n
mean_y = sum(y) / n

# чисельник: сума добутків відхилень (x - x̄)(y - ȳ)
numerator = sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(n))

# знаменник: корінь із добутку двох сум квадратів відхилень
sum_sq_x = sum((xi - mean_x) ** 2 for xi in x)
sum_sq_y = sum((yi - mean_y) ** 2 for  yi in y)
denominator = (sum_sq_x * sum_sq_y) ** 0.5

py_corr = numerator / denominator
print(py_corr)
```

```python
corr_matrix = np.corrcoef(sum_purchases, visits)
np_corr = corr_matrix[0, 1]
print(np_corr)
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
purchases = np.array(sum_purchases)
visits_arr = np.array(visits)

mean_p = np.mean(purchases)
std_p = np.std(purchases)
threshold = 2 * np.std(purchases)

# True для звичайних клієнтів, False для викидів
mask = purchases <= threshold
# залишаємо покупки без викидів
clean_purchases = purchases[mask]
# Залишаємо візитів без викидів
clean_visits = visits_arr[mask]

print(f"2 стандартних відхилення: {threshold}")
print(len(clean_purchases), len(clean_visits))

clean_corr = np.corrcoef(clean_purchases, clean_visits)[0, 1]
print(f"кореляція без викидів: {clean_corr}")

plt.scatter(clean_purchases, clean_visits)
plt.title('Purchases vs. Visits (without outliers)')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```
